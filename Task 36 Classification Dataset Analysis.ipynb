**Objective:** Understand the structure of a basic classification problem by examining features, target variables, and target classes using the Iris dataset.

```python
import pandas as pd

# Load the Iris dataset
df = pd.read_csv('iris_dataset.csv')

# Display the first 5 rows of the dataset
print("--- First 5 Rows of the Dataset ---")
print(df.head())

# Display dataset information
print("\n--- Dataset Info ---")
print(df.info())
```

### Feature and Target Identification
- **Features (Independent Variables):** The measurements used by the model to learn and make predictions. In this dataset, the features are:
  - `sepal length (cm)`
  - `sepal width (cm)`
  - `petal length (cm)`
  - `petal width (cm)`
- **Target Variable (Dependent Variable):** The column we want to predict. Here, the target variable is `species`.

```python
# Check unique values in the target column
target_column = 'species'
unique_classes = df[target_column].unique()

print(f"Target Column: {target_column}")
print(f"Unique Target Classes: {list(unique_classes)}")
print(f"Total Number of Classes: {len(unique_classes)}")

# Determine if binary or multiclass
if len(unique_classes) == 2:
    print("Problem Type: Binary Classification")
else:
    print("Problem Type: Multiclass Classification")
```

### Explanation of the Classification Problem
- **Problem Type:** Multiclass Classification (since there are 3 distinct target classes: `setosa`, `versicolor`, and `virginica`).
- **Goal:** The objective is to train a machine learning algorithm to correctly classify iris flowers into their respective species based on their sepal and petal dimensions.

### Interview Questions & Answers

#### 1. What is classification?
**Answer:** Classification is a type of supervised machine learning where the goal is to predict the categorical class label of a given observation or data point. Examples include spam detection (spam or not spam) and disease diagnosis (positive or negative).

#### 2. What is binary classification?
**Answer:** Binary classification is a classification task where the target variable has exactly two possible classes or outcomes (e.g., Yes/No, True/False, 0/1, Spam/Not Spam). 

#### 3. What is multiclass classification?
**Answer:** Multiclass classification is a classification task where the target variable has three or more possible classes, and each instance is classified into one and only one of those classes (e.g., classifying iris flowers into `setosa`, `versicolor`, or `virginica`).
